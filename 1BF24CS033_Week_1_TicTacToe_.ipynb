```python
class TicTacToe:
    def __init__(self):
        self.board = [" "] * 9

    def print_board(self):
        print("\n")
        for row in range(3):
            print(f" {self.board[row*3]} | {self.board[row*3+1]} | {self.board[row*3+2]} ")
            if row < 2:
                print("---+---+---")
        print("\n")

    def print_board_positions(self):
        print("Board Position Guide:")
        print(" 1 | 2 | 3 ")
        print("---+---+---")
        print(" 4 | 5 | 6 ")
        print("---+---+---")
        print(" 7 | 8 | 9 \n")

    def is_winner(self, board, player):
        win_conditions = [
            [0, 1, 2], [3, 4, 5], [6, 7, 8],  # Rows
            [0, 3, 6], [1, 4, 7], [2, 5, 8],  # Columns
            [0, 4, 8], [2, 4, 6]              # Diagonals
        ]
        return any(all(board[i] == player for i in condition) for condition in win_conditions)

    def is_full(self, board):
        return " " not in board

    def player_move(self, current_player):
        while True:
            try:
                move = int(input(f"Player '{current_player}', enter position (1-9): ")) - 1
                if 0 <= move <= 8:
                    if self.board[move] == " ":
                        self.board[move] = current_player
                        break
                    else:
                        print("Position already taken! Choose an empty spot.")
                else:
                    print("Invalid position! Enter a number between 1 and 9.")
            except ValueError:
                print("Invalid input! Please enter a valid integer.")

    def play(self):
        print("=" * 30)
        print("    WELCOME TO TIC-TAC-TOE    ")
        print("=" * 30 + "\n")

        self.print_board_positions()
        current_player = "X"

        while True:
            self.print_board()
            self.player_move(current_player)

            # Check Win Condition
            if self.is_winner(self.board, current_player):
                self.print_board()
                print(f"🎉 Player '{current_player}' wins! 🎉")
                break

            # Check Tie Condition
            if self.is_full(self.board):
                self.print_board()
                print("It's a draw!")
                break

            # Switch Player between X and O
            current_player = "O" if current_player == "X" else "X"


if __name__ == "__main__":
    while True:
        game = TicTacToe()
        game.play()

        again = input("\nWould you like to play again? (y/n): ").strip().lower()
        if again != "y":
            print("Thanks for playing!")
            break
```

```python
def variable_vacuum_agent():
    # 1. Get total number of rooms
    while True:
        try:
            num_rooms = int(input("Enter total number of rooms: "))
            if num_rooms > 0:
                break
            print("Please enter a number greater than 0.")
        except ValueError:
            print("Invalid input! Please enter an integer.")

    room_names = [f"Room {i+1}" for i in range(num_rooms)]

    # 2. Input dirt status for each room (0 = Clean, 1 = Dirty)
    environment = {}
    print("\nEnter status for each room (0 for Clean, 1 for Dirty):")
    for room in room_names:
        while True:
            status = input(f"  Status of {room}: ").strip()
            if status in ["0", "1"]:
                environment[room] = int(status)
                break
            print("  Invalid status! Enter 0 for Clean or 1 for Dirty.")

    # 3. Choose starting room
    while True:
        print("\nAvailable rooms:", ", ".join(room_names))
        start_input = input("Enter starting room number (e.g., 1): ").strip()
        start_label = f"Room {start_input}"
        if start_label in environment:
            current_idx = room_names.index(start_label)
            break
        print("Invalid room! Selection must be within the available rooms.")

    cost = 0
    direction = 1  # 1 for moving RIGHT, -1 for moving LEFT

    print("\n--- Starting Vacuum Cleaner ---")

    # Sweep through rooms until every room in the environment is clean
    while any(status == 1 for status in environment.values()):
        current_room = room_names[current_idx]

        # Action: Suck if current room is dirty
        if environment[current_room] == 1:
            print(f"Location {current_room} is DIRTY. Action: Suck dirt.")
            environment[current_room] = 0
            cost += 1
        else:
            # Action: Move left or right depending on boundaries
            if current_idx == num_rooms - 1:
                direction = -1  # Turn around at right wall
            elif current_idx == 0:
                direction = 1   # Turn around at left wall

            next_idx = current_idx + direction
            move_str = "RIGHT" if direction == 1 else "LEFT"

            print(f"Location {current_room} is CLEAN. Action: Move {move_str} to {room_names[next_idx]}.")
            current_idx = next_idx
            cost += 1

    print("\n--- Goal State Reached ---")
    print(f"Final Location: {room_names[current_idx]}")
    print("All rooms are clean!")
    print(f"Total actions taken (Cost): {cost}")


if __name__ == "__main__":
    variable_vacuum_agent()
```
